### SERVER & ANALYSIS FUNCTIONS

````python
class SQLManager:
    def __init__(self, db_name='database01.db'):
        """
        Initializes the SQLManager class.

        Args:
            db_name (str): The name of the SQLite database. Defaults to 'test.db'.
        """
        self.engine = create_engine(f'sqlite:///{db_name}', echo=False)

    def start_server(self, df, table_name='Table1'):
        """
        Creates the database and table, and inserts data from the DataFrame.

        Args:
            df (pd.DataFrame): The DataFrame containing the data to insert.
            table_name (str): The name of the table in the database. Defaults to 'Table1'.
        """
        with self.engine.connect() as connection:
            df.to_sql(table_name, connection.connection, index=False, if_exists='replace')

    def execute_query(self, query, return_dataframe=False):
        """
        Executes a SQL query and optionally returns the results as a Pandas DataFrame.

        Args:
            query (str): The SQL query to execute.
            return_dataframe (bool): If True, returns the results as a Pandas DataFrame.
                                     Defaults to False.

        Returns:
            pd.DataFrame or None: The results of the query as a Pandas DataFrame if
                                  return_dataframe is True, otherwise None.
        """
        with self.engine.connect() as connection:
            query_results = connection.execute(text(query)).fetchall()
            df_results = pd.DataFrame(query_results)
            display(df_results)

            if return_dataframe:
                return df_results
            else:
                return None

    def shutdown_server(self):
        """
        Disposes of the SQLAlchemy engine, effectively closing all connections.
        """
        self.engine.dispose()
        print("SQL server has been shut down.")

def plot_countplot(df, col):
    """
    Plots a countplot for a given categorical column in the DataFrame.

    Args:
        df (pd.DataFrame): The DataFrame containing the data.
        col (str): The name of the categorical column to plot.
    """
    if col in df.columns and (isinstance(df[col].dtype, pd.CategoricalDtype) or df[col].dtype == object):
        plt.figure(figsize=(12, 5))
        ax = sns.countplot(data=df, y=col, hue=col, order=df[col].value_counts().index, palette='viridis', legend=False)
        plt.title(f'Countplot of {col}')
        plt.xlabel('Count')
        plt.ylabel(col)
        ### CALCULATE PERCENTAGES
        total = len(df[col])
        value_counts_sorted = df[col].value_counts().reindex(df[col].value_counts().index)
        percentages = value_counts_sorted / total * 100
        percentage_map = percentages.to_dict()

        max_bar_width = ax.get_xlim()[1]
        text_inside_threshold_ratio = 0.1

        ### ADD PERCENTAGE TEXT
        for patch in ax.patches:
            bar_width = patch.get_width()
            patch_center_y = patch.get_y() + patch.get_height() / 2

            yticks = ax.get_yticks()
            yticklabels = ax.get_yticklabels()
            closest_tick_index = (abs(yticks - patch_center_y)).argmin()
            category_label = yticklabels[closest_tick_index].get_text()

            percentage = percentage_map.get(category_label, 0.0)
            text = f'{percentage:.1f}%'

            if bar_width > max_bar_width * text_inside_threshold_ratio:
                ### PLACE TEXT INSIDE BAR
                x_position = bar_width - 10
                color = 'white'
                ha = 'right'
            else:
                ### PLACE TEXT OUTSIDE BAR
                x_position = bar_width + 5
                color = 'black'
                ha = 'left'

            ax.text(x_position,
                    patch_center_y,
                    text,
                    va='center',
                    ha=ha,
                    color=color,
                    fontweight='bold')
        ### ADD LEGEND
        plt.tight_layout()
        plt.show()
    elif col not in df.columns:
        print(f"Warning: Column '{col}' not found in DataFrame. Skipping.")
    else:
        print(f"Warning: Column '{col}' is not categorical. Skipping plotting.")
````

````python
### EVALUATING THE DEPENDENCY OF CATEGORICAL FEATURES
categorical_cols_for_analysis = df.select_dtypes(include=['category','object']).columns.tolist()
results = []

# CHI-SQUARED TEST FOR DEPENDENCY
for i in range(len(categorical_cols_for_analysis)):
    for j in range(i + 1, len(categorical_cols_for_analysis)):
        col1 = categorical_cols_for_analysis[i]
        col2 = categorical_cols_for_analysis[j]

        crosstab = pd.crosstab(df[col1], df[col2])

        chi2, p, dof, expected = chi2_contingency(crosstab)

        # P-VALUE AND ALPHA DECISION
        alpha = 0.05
        decision = 'Reject H0 (Dependent)' if p < alpha else 'Fail to Reject H0 (Independent)'

        results.append({
            'Variable Pair': f'{col1} vs {col2}',
            'Chi-squared Statistic': chi2,
            'P-value': p,
            'Decision': decision
        })

results_df = pd.DataFrame(results)

print(tabulate(results_df, headers='keys', tablefmt='grid'))
````

When combined with their shared strong associations with pain intensity and migraine type classification, these core pain descriptors form what appears to be a diagnostic constellation that likely corresponds to different trigeminovascular activation patterns documented in migraine pathophysiology literature.

The strong association between nausea and photophobia (χ² = 58.582, p < 0.001) while nausea remains independent of phonophobia indicates nuanced differences in how these sensory pathways interact with autonomic centers.

Vertigo emerges as a central connector, significantly associating with multiple symptoms including tinnitus (χ² = 43.766, p < 0.001), visual defects, and paresthesia, suggesting vestibular involvement creates a cascade of related neurological symptoms. This is especially true in conditions like migraines, where a wide variety of overlapping symptoms can make accurate classification challenging. Machine learning (ML) offers a solution by leveraging algorithms that learn from data patterns, making it possible to uncover subtle associations that might not be evident through manual analysis alone.

- In the context of migraine diagnosis, ML models can support clinicians by providing consistent, data-driven classifications based on symptom profiles. This is particularly valuable when differentiating between migraine subtypes, which may require different treatments or have distinct prognoses. By automating the recognition of these patterns, ML can reduce diagnostic ambiguity, improve early detection, and enhance the personalization of care.

- One of the key strengths of ML lies in its ability to handle large sets of features and find non-linear interactions among them. This makes it ideal for medical datasets where many variables — such as sensory disturbances, visual symptoms, and family history — may all interact to shape a diagnosis. Additionally, ML models can be retrained and refined as more data becomes available, continually improving their performance over time.

- Given the imbalance in the distribution of migraine types, with some categories being underrepresented, it was necessary to apply a balancing technique to ensure fair and meaningful learning.
````python
## FINAL REVISION
if df.isnull().sum().sum() > 0:
    print("Warning: Missing values found. Implement handling strategy.")
else:
    print("No missing values found.")
````

````python
#### DATA TRANSFORMATION ####
### ENCODING CATEGORICAL VARIABLES

## MAP CATEGORICAL VARIABLES TO NUMERICAL (ACCORDING TO THE DATA DICTIONARY)
location_map_new = {'None': 0, 'Unilateral': 1, 'Bilateral': 2}
character_map_new = {'None': 0, 'Throbbing': 1, 'Constant': 2}
intensity_map_new = {'None': 0, 'Mild': 1, 'Medium': 2, 'Severe': 3}
binary_map_new = {'Not': 0, 'Yes': 1}

df_processed = df.copy()

# MAPPING CATEGORICAL VARIABLES
categorical_cols_to_encode = [
    'Location', 'Character', 'Intensity', 'Nausea', 'Vomit',
    'Phonophobia', 'Photophobia', 'Dysphasia', 'Dysarthria', 'Vertigo',
    'Tinnitus', 'Hypoacusis', 'Diplopia', 'Visual_Defect', 'Ataxia',
    'Conscience', 'Paresthesia', 'DPF'
]

for col in categorical_cols_to_encode:
    if col in df_processed.columns and isinstance(df_processed[col].dtype, pd.CategoricalDtype):
        if col in ['Location', 'Character', 'Intensity']:
             # Use specific maps for these columns
             if col == 'Location':
                 df_processed[col] = df_processed[col].cat.rename_categories(location_map_new)
             elif col == 'Character':
                 df_processed[col] = df_processed[col].cat.rename_categories(character_map_new)
             elif col == 'Intensity':
                 df_processed[col] = df_processed[col].cat.rename_categories(intensity_map_new)
        else:
            # Use binary map for other categorical columns
            df_processed[col] = df_processed[col].cat.rename_categories(binary_map_new)

        df_processed[col] = df_processed[col].astype(int)
    elif col not in df_processed.columns:
         print(f"Warning: Column '{col}' not found in the DataFrame during encoding.")
    else:
        print(f"Warning: Column '{col}' is not categorical. Skipping encoding.")
````

````python
## SEPARATE FEATURES (X) AND TARGET (y)
X = df_processed.drop('Type', axis=1)
y = df_processed['Type']

## ENCODE TARGET VARIABLE (Type)
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
````

````python
## SPLIT THE DATASET (BEFORE SMOTE)
X_train, X_test, y_train_encoded, y_test_encoded = train_test_split(
    X, y_encoded, test_size=0.3, random_state=123, stratify=y_encoded
)

print("Original dataset shape %s" % Counter(y_encoded))
print("\nDataset split into training and testing sets (before SMOTE).")
print(f"Training set shape (X_train, y_train_encoded): {X_train.shape}, {y_train_encoded.shape}")
print(f"Testing set shape (X_test, y_test_encoded): {X_test.shape}, {y_test_encoded.shape}")
print("Training set distribution %s" % Counter(y_train_encoded))
print("Testing set distribution %s" % Counter(y_test_encoded))
````

````python
## SCALING NUMERICAL VARIABLES
numerical_cols = X.select_dtypes(include=['int64', 'float64', 'int32', 'float32']).columns
scaler = StandardScaler()

X_train[numerical_cols] = scaler.fit_transform(X_train[numerical_cols])
X_test[numerical_cols] = scaler.transform(X_test[numerical_cols])
````

````python
## APPLICATION OF SMOTE
smote = SMOTE(random_state=999)
X_train_resampled, y_train_resampled_encoded = smote.fit_resample(X_train, y_train_encoded)

print("\nApplying SMOTE to the training data...")
print("Resampled training dataset shape %s" % Counter(y_train_resampled_encoded))
````

````python
## MODEL SELECTION FOR COMPARISON
models = {
    'Logistic Regression': LogisticRegression(random_state=999, solver='liblinear'),
    'Decision Tree': DecisionTreeClassifier(random_state=999),
    'XGBoost': XGBClassifier(random_state=999, eval_metric='mlogloss'),
    'SVM': SVC(random_state=999, probability=True),
    'Random Forest': RandomForestClassifier(random_state=999)
}
````

````python
## START TRAINING AND EVALUATION
print("--- Cross-validation Results on Training Data ---")
cv_results = {}
for name, model in models.items():
    cv_scores = cross_val_score(model, X_train_resampled, y_train_resampled_encoded, cv=10)
    cv_results[name] = cv_scores
    print(f"{name} Cross-validation Accuracy: {cv_scores.mean():.4f} (+/- {cv_scores.std():.4f})")

print("-" * 40)
````

````python
## BEST MODEL SELECTION
mean_cv_scores = {name: scores.mean() for name, scores in cv_results.items()}
best_cv_model_name = max(mean_cv_scores, key=mean_cv_scores.get)
print(f"Model with best cross-validation score on training data: {best_cv_model_name}")
print("-" * 40)
````

````python
## EVALUATION ON HELD-OUT TEST DATA
print("--- Evaluation on Held-out Test Data ---")
evaluation_results = []
for name, model in models.items():
    print(f"Evaluating {name}...")
    ## COMPLETE TRAINING
    model.fit(X_train_resampled, y_train_resampled_encoded)
    # PREDICT ON TEST DATA
    y_pred_encoded = model.predict(X_test)
    # INVERSE TRANSFORM PREDICTIONS
    y_pred = label_encoder.inverse_transform(y_pred_encoded)
    y_test_original = label_encoder.inverse_transform(y_test_encoded)

    accuracy = accuracy_score(y_test_original, y_pred)
    report = classification_report(y_test_original, y_pred, output_dict=True)

    evaluation_results.append({
        'Model': name,
        'Accuracy': accuracy,
        'Report': report
    })

    ## DETAILED REPORT
    print(f"\nResults for {name}:")
    print(f"Accuracy: {accuracy:.4f}")
    print("Classification Report:")
    print(classification_report(y_test_original, y_pred))
    print("-" * 30)

print("-" * 40)

evaluation_df = pd.DataFrame(evaluation_results)
````

````python
### PLOT MODEL PERFORMANCE
print("--- Visualizing Model Performance on Test Data ---")

# ACCURACY COMPARISON
plt.figure(figsize=(10, 6))
ax = sns.barplot(x='Model', y='Accuracy', data=evaluation_df, palette='viridis')
plt.title('Model Accuracy Comparison on Test Data')
plt.ylabel('Accuracy')
plt.ylim(0, 1)
plt.xticks(rotation=45, ha='right')

for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{height:.2f}',
                (p.get_x() + p.get_width() / 2., height / 2),  # position inside bar vertically centered
                ha='center',
                va='center',
                fontsize=12,
                color='white',
                rotation=90)

plt.tight_layout()
plt.show()

# F1-SCORE COMPARISON
f1_scores_list = []
for index, row in evaluation_df.iterrows():
    model_name = row['Model']
    report = row['Report']
    for class_label, metrics in report.items():
        if class_label not in ['accuracy', 'macro avg', 'weighted avg']:
            f1_scores_list.append({'Model': model_name, 'Class': class_label, 'F1-score': metrics['f1-score']})

f1_scores_df = pd.DataFrame(f1_scores_list)

plt.figure(figsize=(14, 7))
ax = sns.barplot(x='Class', y='F1-score', hue='Model', data=f1_scores_df, palette='viridis')
plt.title('F1-score Comparison per Class on Test Data')
plt.xlabel('Migraine Type')
plt.ylabel('F1-score')
plt.ylim(0, 1)
plt.xticks(rotation=45, ha='right')

for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{height:.2f}',
                (p.get_x() + p.get_width() / 2., height / 2),
                ha='center',
                va='center',
                fontsize=12,
                color='white',
                rotation=90)

ax.legend(loc='upper left', bbox_to_anchor=(1, 1))

plt.tight_layout()
plt.subplots_adjust(right=0.85)
plt.show()
````

````python
## BEST MODEL CONFUSION MATRIX
best_test_model_row = evaluation_df.loc[evaluation_df['Accuracy'].idxmax()]
best_test_model_name = best_test_model_row['Model']
best_test_model = models[best_test_model_name]

y_pred_best_test_encoded = best_test_model.predict(X_test)
y_pred_best_test = label_encoder.inverse_transform(y_pred_best_test_encoded)
y_test_original_for_plot = label_encoder.inverse_transform(y_test_encoded)

print(f"--- Confusion Matrix for the Best Model on Test Data ({best_test_model_name}) ---")
cm_best_test = confusion_matrix(y_test_original_for_plot, y_pred_best_test)

plt.figure(figsize=(10, 8))
disp = ConfusionMatrixDisplay(confusion_matrix=cm_best_test, display_labels=label_encoder.classes_)
disp.plot(cmap=plt.cm.Blues, values_format='d')
plt.title(f'Confusion Matrix - {best_test_model_name} (Test Data)')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.xticks(rotation=45, ha='right')
plt.show()
````

### (Additonal) MODEL PERFORMANCE WITH SMOTE (BEFORE SPLIT) - TARGET LEAK
---

- In machine learning, particularly when addressing class imbalance with techniques like SMOTE (Synthetic Minority Over-sampling Technique), it's crucial to apply these methods correctly to avoid data leakage. Data leakage occurs when information from the test set inadvertently influences the training process, leading to overly optimistic performance estimates that don't generalize well to unseen data. Applying SMOTE before splitting the dataset into training and test sets can cause synthetic samples to be influenced by the entire dataset, including the test set, thereby introducing leakage. This practice can result in inflated performance metrics, as the model may inadvertently learn from data it should not have access to during training.

- To mitigate this risk, it's recommended to first split the dataset into training and test sets, and then apply SMOTE exclusively to the training data. This approach ensures that the model learns only from the training data, preserving the integrity of the test set for unbiased evaluation. By doing so, the synthetic samples generated by SMOTE are based solely on the training data, preventing any information from the test set from leaking into the training process .

````python
#### REPRETING THE SAME PRE-PROCESSING ####

### ENCODING CATEGORICAL VARIABLES

## MAP CATEGORICAL VARIABLES TO NUMERICAL (ACCORDING TO THE DATA DICTIONARY)
location_map_new = {'None': 0, 'Unilateral': 1, 'Bilateral': 2}
character_map_new = {'None': 0, 'Throbbing': 1, 'Constant': 2}
intensity_map_new = {'None': 0, 'Mild': 1, 'Medium': 2, 'Severe': 3}
binary_map_new = {'Not': 0, 'Yes': 1}

df_processed = df.copy()

# MAPPING CATEGORICAL VARIABLES
categorical_cols_to_encode = [
    'Location', 'Character', 'Intensity', 'Nausea', 'Vomit',
    'Phonophobia', 'Photophobia', 'Dysphasia', 'Dysarthria', 'Vertigo',
    'Tinnitus', 'Hypoacusis', 'Diplopia', 'Visual_Defect', 'Ataxia',
    'Conscience', 'Paresthesia', 'DPF'
]

for col in categorical_cols_to_encode:
    if col in df_processed.columns and isinstance(df_processed[col].dtype, pd.CategoricalDtype):
        if col in ['Location', 'Character', 'Intensity']:
             # Use specific maps for these columns
             if col == 'Location':
                 df_processed[col] = df_processed[col].cat.rename_categories(location_map_new)
             elif col == 'Character':
                 df_processed[col] = df_processed[col].cat.rename_categories(character_map_new)
             elif col == 'Intensity':
                 df_processed[col] = df_processed[col].cat.rename_categories(intensity_map_new)
        else:
            # Use binary map for other categorical columns
            df_processed[col] = df_processed[col].cat.rename_categories(binary_map_new)

        df_processed[col] = df_processed[col].astype(int)
    elif col not in df_processed.columns:
         print(f"Warning: Column '{col}' not found in the DataFrame during encoding.")
    else:
        print(f"Warning: Column '{col}' is not categorical. Skipping encoding.")

## SEPARATE FEATURES (X) AND TARGET (y)
X = df_processed.drop('Type', axis=1)
y = df_processed['Type']

## ENCODE TARGET VARIABLE (Type)
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

#### CHANGE #####

## SCALING NUMERICAL VARIABLES (BEFORE SPLIT)
scaler = StandardScaler()
X[numerical_cols] = scaler.fit_transform(X[numerical_cols])

## APPLICATION OF SMOTE
smote = SMOTE(random_state=123)
X_resampled, y_resampled_encoded = smote.fit_resample(X, y_encoded)

#### END CHANGE ####

print("Original dataset shape %s" % Counter(y_encoded))
print("Resampled dataset shape %s" % Counter(y_resampled_encoded))

## SPLIT THE DATASET (AFTER SMOTE)
X_train, X_test, y_train_encoded, y_test_encoded = train_test_split(
    X_resampled, y_resampled_encoded, test_size=0.3, random_state=123, stratify=y_resampled_encoded
)

print("\nDataset split into training and testing sets.")
print(f"Training set shape (X_train, y_train_encoded): {X_train.shape}, {y_train_encoded.shape}")
print(f"Testing set shape (X_test, y_test_encoded): {X_test.shape}, {y_test_encoded.shape}")
````

````python
##### MODELLING PHASE (NO CHANGES) #####

## MODEL SELECTION FOR COMPARISON
models = {
    'Logistic Regression': LogisticRegression(random_state=999, solver='liblinear'),
    'Decision Tree': DecisionTreeClassifier(random_state=999),
    'XGBoost': XGBClassifier(random_state=999, eval_metric='mlogloss'),
    'SVM': SVC(random_state=999, probability=True),
    'Random Forest': RandomForestClassifier(random_state=999)
}

## START TRAINING AND EVALUATION
print("--- Cross-validation Results on Training Data ---")
cv_results = {}
for name, model in models.items():
    cv_scores = cross_val_score(model, X_train, y_train_encoded, cv=5)
    cv_results[name] = cv_scores
    print(f"{name} Cross-validation Accuracy: {cv_scores.mean():.4f} (+/- {cv_scores.std():.4f})")

print("-" * 40)

## BEST MODEL SELECTION
mean_cv_scores = {name: scores.mean() for name, scores in cv_results.items()}
best_cv_model_name = max(mean_cv_scores, key=mean_cv_scores.get)
print(f"Model with best cross-validation score on training data: {best_cv_model_name}")
print("-" * 40)

for name, model in models.items(): 
    print(f"Evaluating {name}...")
    ## COMPLETE TRAINNING
    model.fit(X_train, y_train_encoded) ## FITTING NEW VALUES OF X AND Y 
    # PREDICT ON TEST DATA
    y_pred_encoded = model.predict(X_test)
    # INVERSE TRANSFORM PREDICTION
    y_pred = label_encoder.inverse_transform(y_pred_encoded)
    y_test_original = label_encoder.inverse_transform(y_test_encoded)

    accuracy = accuracy_score(y_test_original, y_pred)
    report = classification_report(y_test_original, y_pred, output_dict=True)

    evaluation_results.append({
        'Model': name,
        'Accuracy': accuracy,
        'Report': report
    })

    ## DETAILED REPORT
    print(f"\nResults for {name}:")
    print(f"Accuracy: {accuracy:.4f}")
    print("Classification Report:")
    print(classification_report(y_test_original, y_pred))
    print("-" * 30)

print("-" * 40)

evaluation_df = pd.DataFrame(evaluation_results)
````
